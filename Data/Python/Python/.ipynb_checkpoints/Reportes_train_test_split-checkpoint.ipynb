```python
import concurrent.futures
import pandas as pd
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import numpy
from sklearn import metrics
```

```python
def MatrizConfusion(y_test, y_pred):
            matriz_confusion = confusion_matrix(y_test, y_pred)
            print(matriz_confusion)
            # Crear la visualización de la matriz de confusión
            cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix = matriz_confusion, display_labels = [1, 0])
            # Graficar la matriz de confusión
            cm_display.plot()
            plt.show()
```

```python
from sklearn.metrics import recall_score
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import f1_score

def ExecuteIteration(model,train_idx, test_idx,X,y):
    # Training fold
    x_train = X.iloc[train_idx]
    y_train = y[train_idx]
    # Testing fold
    x_test = X.iloc[test_idx]
    y_test = y[test_idx]
    
    print('Train X ',x_train.shape, ' Train y ',y_train.shape)
    print('Test X ',x_test.shape, ' Test y ',y_test.shape)
    print("Entrenamiento: Conteo: ",y_train.value_counts())
    print("Prueba: Conteo: ",y_test.value_counts())
    
#     x_train,y_train = class_balance("SMOTE",x_train,y_train)
    # Train
    model.fit(x_train, y_train)

    # Get metrics
    predict = model.predict(x_test)
    MatrizConfusion(y_test,predict)
    
    accuracy = accuracy_score(y_test, predict)
    precision = precision_score(y_test, predict)
    recall = recall_score(y_test, predict)
    f1score = f1_score(y_test, predict)
    
#     print('Acurrency: ',accuracy,' Precision: ',precision,' Recall: ',recall,' F1:',f1score)
    
    return accuracy,precision,recall,f1score
```

```python
# Metrics
import numpy as np
from joblib import Parallel, delayed
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import StratifiedKFold

def classify(X, y,model):
    print(X.shape)
    skf = StratifiedKFold(n_splits=10)
    skf.get_n_splits(X, y)
    for i, (train_index, test_index) in enumerate(skf.split(X, y)):
        print(train_index.shape , test_index.shape )
        print(ExecuteIteration(model,train_index, test_index,X,y))
    
    metrics_df = pd.DataFrame({
        'Results': [
            round(0, 2),
            round(0, 2),
            round(0, 2),
            round(0, 2)
        ]
    }, index=["pr", "re", "f", "Ac"])

    return metrics_df
```

```python
from sklearn.naive_bayes import GaussianNB
def NaiveBayes():
    # Create the Gaussian Naive Bayes model
    return GaussianNB()
```

```python
from sklearn.tree import DecisionTreeClassifier
def DecisionTree():
    # Create the DecisionTreeClassifier model
    return DecisionTreeClassifier()
```

```python
from sklearn.ensemble import RandomForestClassifier
def RandomForest():
    # Create the RandomForestClassifier model
    return RandomForestClassifier(
    n_estimators=100,  # Número de árboles en el bosque
    criterion="gini",  # Criterio de división (puede ser "entropy" también)
    max_depth=None,    # Profundidad máxima de los árboles (ajusta según tus necesidades)
    min_samples_split=2,  # Número mínimo de muestras requeridas para dividir un nodo
    min_samples_leaf=1,   # Número mínimo de muestras requeridas en una hoja
    random_state=42     # Semilla aleatoria para reproducibilidad
)
```

```python
from imblearn.over_sampling import RandomOverSampler
from imblearn.under_sampling import RandomUnderSampler
from sklearn.model_selection import train_test_split

def ExecuteModel(model_name,X,y):
    try:
        # Algoritmo a ejecutar
        if model_name == "NB":
          return classify(X, y,NaiveBayes()) 
        elif model_name == "DT":
          return classify(X, y,DecisionTree()) 
        elif model_name == "RF":
          return classify(X, y, RandomForest())
    except Exception as e:
     print(str(e))
     return pd.DataFrame([-1,-1,-1,-1],
                    index=["pr","re","f","Ac"],
                    columns=['Result'])
```

```python
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import RandomOverSampler
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import SMOTE
def class_balance(IntanceReduction,X,y):  
    if IntanceReduction == "RUS":
       rus = RandomUnderSampler(random_state=42,sampling_strategy='majority')
       X,y = rus.fit_resample(X,y)
    elif IntanceReduction == "ROS":
       ros = RandomOverSampler(random_state=42,sampling_strategy='minority')
       X,y = ros.fit_resample(X,y)
    elif IntanceReduction == "SMOTE":
        smote = SMOTE()
        X,y = smote.fit_resample(X,y)
    return X,y
```

```python
from sklearn.preprocessing import StandardScaler
def get_data(path):
    data=pd.read_csv(path)
    data = data.dropna()
    data=data.drop(columns=['Project','Hash', 'LongName','Parent'],axis=1, errors='ignore')
    data['Number of Bugs'] = data['Number of Bugs'].map(lambda x: 1 if x > 0 else 0)
    X =data.drop(labels=['Number of Bugs'], axis=1).astype(float).abs()
    y = data['Number of Bugs'] 
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    X = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
#     return class_balance('SMOTE',X,y)
    return X,y
```

```python
def remove_features(X,project,b,rank,level,modelsSelection):
    dataMetric = metricsList.loc[(metricsList['project'] == project) & (metricsList['level_measure'] == level) & (metricsList['b'] == b) & (metricsList['rank'] == rank)] 
    dataMetric = dataMetric.loc[dataMetric[modelsSelection] == 'True']
    dataMetric = dataMetric['metric'].values  
    # Elimina las columnas del DataFrame que no están en column_array
    X = X.drop(columns=[col for col in X.columns if col not in dataMetric])
    return X
```
