```python
import pandas as pd
import numpy as np
```

```python
import os

def clear_console():
    if os.name == 'posix':  # Unix/Linux/MacOS
        os.system('clear')
    elif os.name == 'nt':   # Windows
        os.system('cls')
```

```python
import os
ruta_actual =os.path.abspath(os.getcwd()) 

folder_selection =os.path.join(ruta_actual, 'selection') 
folder_dataset =os.path.join(ruta_actual, 'dataset')
folder_result =os.path.join(ruta_actual, 'results')
```

```python
class ModelResult:
    def __init__(self, level_measure,project,b,rank,model_selection,intance_reduction,model,precision,recall,f,accuracy,count_metric,count_row,count_bag):
        self.level_measure = level_measure
        self.project = project
        self.b = b
        self.rank = rank
        self.model_selection = model_selection
        self.intance_reduction = intance_reduction
        self.model = model
        self.precision = precision
        self.recall = recall
        self.f = f
        self.accuracy = accuracy
        self.count_metric = count_metric
        self.count_row = count_row
        self.count_bag = count_bag
class SelectMeasure:
    def __init__(self,level_measure,project,b,rank,cs_su,ig_su,rf_su,cs_cos,ig_cos,rf_cos):
        self.level_measure = level_measure
        self.project = project
        self.b = b
        self.rank = rank
        self.cs_su = cs_su
        self.ig_su = ig_su
        self.rf_su = rf_su
        self.cs_cos = cs_cos
        self.ig_cos = ig_cos
        self.rf_cos = rf_cos
```

```python
import os
def FileList(directory,list=[]):
  for filename in os.listdir(directory):
      f = os.path.join(directory, filename)
      if os.path.isdir(f):
        FileList(f,list)
      # checking if it is a file
      if os.path.isfile(f) and os.path.splitext(f)[-1].lower() == '.csv':
        list.append(f)
  return np.array(list)
```

```python
def SelectAllMetricsDb():
    return pd.read_csv(folder_selection+'/select_measure.csv', dtype=str)
```

```python
listFile = FileList(folder_dataset)
projects =np.unique([os.path.split(os.path.dirname(i))[1] for i in listFile])
```

```python
metricsList = SelectAllMetricsDb().dropna(subset=['metric'])
model_results = []
```

```python
from models.modelsmote import *
```

```python
from sklearn.preprocessing import StandardScaler
def ModelsByCondition(project):
        model_project = []
        ranks = ['0.5','0.7','0.9']
        bs = ['0.5','0.7','0.9']
        models = ['RF','NB','DT']
        modelsSelections = ['ALL','cs_su','ig_su','rf_su','cs_cos','ig_cos','rf_cos']
        IntanceReductions = ['NONE']
        for rank in ranks:
              for b in bs:
                    paths = [i for i in listFile if os.path.split(os.path.dirname(i))[1] == project] 
                    for path in paths:
                        print(rank,'--',b,'--',path)
                        data=pd.read_csv(path)
                        data = data.dropna()
                        try:
                          data=data.drop(columns=['Project','Hash', 'LongName','Parent'],axis=1, errors='ignore')
                          data['Number of Bugs'] = data['Number of Bugs'].map(lambda x: 1 if x > 0 else 0)
                          X =data.drop(labels=['Number of Bugs'], axis=1).astype(float).abs()
                          y = data['Number of Bugs'] 
                          
                        except Exception as e:
                          print("Error in " + file_location + str(e))
                          continue
                        for model in models:
                            for modelsSelection in modelsSelections:
                                for reduction in IntanceReductions:
                                  level = os.path.split(os.path.basename(path))[1].split('.')[0] 
                                  countBug = sum(1 for i in y if i > 0)
                                  if modelsSelection != 'ALL':
                                      dataMetric = metricsList.loc[(metricsList['project'] == project) & (metricsList['level_measure'] == level) & (metricsList['b'] == b) & (metricsList['rank'] == rank)] 
                                      dataMetric = dataMetric.loc[dataMetric[modelsSelection] == 'True']
                                      dataMetric = dataMetric['metric'].values  
                                      # Elimina las columnas del DataFrame que no están en column_array
                                      X = X.drop(columns=[col for col in X.columns if col not in dataMetric])

                                  if X.shape[1] > 0:
                                      X_data,y_data = GetData(reduction,X,y)
                                      result = ExecuteModel(model,X_data,y_data)
                                      result_class = ModelResult(level,project,b,rank,modelsSelection,reduction,model,result['Results']['pr'],result['Results']['re'],result['Results']['f'],result['Results']['Ac'],X_data.shape[1],X_data.shape[0],countBug)
                                      model_results.append(result_class)
                                      model_project.append(result_class)
        return pd.DataFrame([vars(mr) for mr in model_project])
        
```

```python
for project in projects:
    clear_console()
    print(project,'-------------------------------------------------------------------------------')
    result = ModelsByCondition(project)
    result.to_excel(os.path.join(folder_result,'none', project+'.xlsx'), index=True)
```
