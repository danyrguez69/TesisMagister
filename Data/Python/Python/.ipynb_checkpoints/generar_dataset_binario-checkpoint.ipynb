```python
import pandas as pd
import numpy as np
```

```python
import os
ruta_actual =os.path.abspath(os.getcwd()) 
folder_dataset =os.path.join(ruta_actual, 'dataset_claseBinaria')
```

```python
import os
def FileList(directory,list=[]):
  for filename in os.listdir(directory):
      f = os.path.join(directory, filename)
      if os.path.isdir(f):
        FileList(f,list)
      # checking if it is a file
      if os.path.isfile(f) and os.path.splitext(f)[-1].lower() == '.csv':
        list.append(f)
  return np.array(list)
```

```python
listFile = FileList(folder_dataset)
projects =np.unique([os.path.split(os.path.dirname(i))[1] for i in listFile])
print(projects)
```

```python
for project in projects:
    paths = [i for i in listFile if os.path.split(os.path.dirname(i))[1] == project] 
    for path in paths:
        data=pd.read_csv(path)
        data=data.drop(columns=['Project','Hash', 'LongName','Parent'],axis=1, errors='ignore')
        data['Number of Bugs'] = data['Number of Bugs'].map(lambda x: 'Bug' if x > 0 else 'NoBug')
        data.to_csv(path, index=False)
```
