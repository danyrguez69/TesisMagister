```python
import pandas as pd
import numpy as np
```

```python
import os
ruta_actual =os.path.abspath(os.getcwd()) 
folder_dataset =os.path.join(ruta_actual, 'dataset_weka')
```

```python
import os
def FileList(directory,list=[]):
  for filename in os.listdir(directory):
      f = os.path.join(directory, filename)
      if os.path.isdir(f):
        FileList(f,list)
      # checking if it is a file
      if os.path.isfile(f) and os.path.splitext(f)[-1].lower() == '.csv':
        list.append(f)
  return np.array(list)
```

```python
listFile = FileList(folder_dataset)
projects =np.unique([os.path.split(os.path.dirname(i))[1] for i in listFile])
```

```python
from models.balanceoentrenamiento import *
```

```python
def ModelsByCondition(project):
        model_project = []
        ranks = ['0.5','0.7','0.9']
        bs = ['0.5','0.7','0.9']
        modelsSelections = ['cs_su','ig_su','rf_su','cs_cos','ig_cos','rf_cos']
        IntanceReductions = ['SMOTE','ROS','RUS','NONE']
        paths = [i for i in listFile if os.path.split(os.path.dirname(i))[1] == project]
        for rank in ranks:
              for b in bs:
                    for path in paths:
                            for modelsSelection in modelsSelections:
                                for reduction in IntanceReductions:
                                    try:
                                        level = os.path.split(os.path.basename(path))[1].split('.')[0]
                                        X,y = get_data(path)
                                        X = remove_features(X,project,b,rank,level,modelsSelection)
                                        df_X = pd.DataFrame(X)
                                        df_y = pd.DataFrame(y)

                                        # Unir las características y las clases en un único DataFrame
                                        df_completo = pd.concat([df_X, df_y], axis=1)
                                        folder = os.path.split(os.path.basename(path))[0]
                                        folder = os.path.join(folder, rank+'_'+b+'_'+modelsSelection+'_'+reduction+'_'+level+'.csv') 
                                        df_completo['Number of Bugs'] = df_completo['Number of Bugs'].map(lambda x: 'Bug' if x > 0 else 'NoBug')
                                        df_completo.to_csv(folder, index=False)
                                    except Exception as e:
                                      print("Error in ", str(e))
                                      continue
```

```python
for project in projects:
    ModelsByCondition(project)
#     paths = [i for i in listFile if os.path.split(os.path.dirname(i))[1] == project] 
#     for path in paths:
#         data=pd.read_csv(path)
#         data=data.drop(columns=['Project','Hash', 'LongName','Parent'],axis=1, errors='ignore')
#         data['Number of Bugs'] = data['Number of Bugs'].map(lambda x: 'Bug' if x > 0 else 'NoBug')
#         data.to_csv(path, index=False)
```
