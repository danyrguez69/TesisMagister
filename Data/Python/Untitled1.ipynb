```python
class SelectMeasure:
    def __init__(self, metric,level_measure,project,b,rank,cs_su,ig_su,rf_su,cs_cos,ig_cos,rf_cos):
        self.metric = metric
        self.level_measure = level_measure
        self.project = project
        self.b = b
        self.rank = rank
        self.cs_su = cs_su
        self.ig_su = ig_su
        self.rf_su = rf_su
        self.cs_cos = cs_cos
        self.ig_cos = ig_cos
        self.rf_cos = rf_cos
```

```python
import pyodbc 
def insertDb(row):
    conn = pyodbc.connect('Driver={SQL Server};'
                      'Server=JORDANYS;'
                      'Database=magister;'
                      'Trusted_Connection=yes;')
    cursor = conn.cursor()
    data = "'"+ row.metric+"',"+"'"+ row.level_measure+"',"+"'"+ row.project+"',"+"'"+ row.b+"',"+"'"+ row.rank+"',"+"'"+ row.cs_su+"',"+"'"+ row.ig_su+"',"+"'"+ row.rf_su+"',"+"'"+ row.cs_cos+"',"+"'"+ row.ig_cos+"',"+"'"+ row.rf_cos+"'"
    print(data)
    cursor.execute('''
                INSERT INTO select_measure ( [metric],[level_measure],[project],[b],[rank],[cs_su],[ig_su],[rf_su],[cs_cos],[ig_cos],[rf_cos])
                VALUES
                ('''+ data+''')
               
                ''')

    conn.commit()
    
```

```python
row = SelectMeasure('CC','class','proj 1','0.5','0.8','1','1','1','1','1','1')
insertDb(row)
```
