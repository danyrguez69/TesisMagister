```python
class Pair:
    def __init__(self, pairList,value):
        self.pairList = pairList
        self.value = value
class Features:
    def __init__(self, name,pairs = [],nameJ=[]):
        self.name = name
        self.pairs = pairs
        self.nameJ = nameJ
class SelectMeasure:
    def __init__(self, metric,level_measure,project,b,rank,cs_su,ig_su,rf_su,cs_cos,ig_cos,rf_cos):
        self.metric = metric
        self.level_measure = level_measure
        self.project = project
        self.b = b
        self.rank = rank
        self.cs_su = cs_su
        self.ig_su = ig_su
        self.rf_su = rf_su
        self.cs_cos = cs_cos
        self.ig_cos = ig_cos
        self.rf_cos = rf_cos
```

```python
# from scipy.stats import entropy

def SuPair(X,y):
  try:
    entropyX = round(entropy(X[X.columns[0]]),5)
    entropyY = round(entropy(y[y.columns[0]]),5)
    sumEntropy = entropyX + entropyY
    gainValue = gain(X,y)
    su = 0
    if sumEntropy != 0.0 and gainValue != 0.0:
      su =  round(2 * (gainValue/sumEntropy),5)
    return  Pair([X.columns[0],y.columns[0]],su)
  except Exception as e: print(e)
```

```python
##Entropy

def entropy(Y):
    """
    Also known as Shanon Entropy
    Reference: https://en.wikipedia.org/wiki/Entropy_(information_theory)
    """
    unique, count = np.unique(Y, return_counts=True, axis=0)
    prob = count/len(Y)
    en = np.sum((-1)*prob*np.log2(prob))
    return en


#Joint Entropy

def jEntropy(Y,X):
    """
    H(Y;X)
    Reference: https://en.wikipedia.org/wiki/Joint_entropy
    """
    YX = np.c_[Y,X]
    return entropy(YX)

#Conditional Entropy

def cEntropy(Y, X):
    """
    conditional entropy = Joint Entropy - Entropy of X
    H(Y|X) = H(Y;X) - H(X)
    Reference: https://en.wikipedia.org/wiki/Conditional_entropy
    """
    return jEntropy(Y, X) - entropy(X)


#Information Gain

def gain(Y, X):
    """
    Information Gain, I(Y;X) = H(Y) - H(Y|X)
    Reference: https://en.wikipedia.org/wiki/Information_gain_in_decision_trees#Formal_definition
    """

    return  entropy(Y) - cEntropy(Y,X)
```

```python
from scipy import spatial

def Cos(X,y,name):
  try:
    results = []
    for column in X.columns:
        result = 1 - spatial.distance.cosine(X[column], y)
        results.append(result)
        
    return pd.DataFrame(results,
                    index=X.columns,
                    columns=[name+"COS"]) 
  except:
     return pd.DataFrame([],
                    index=X.columns,
                    columns=[name+"COS"])  
```

```python
from scipy import spatial

def CosPair(X,y):
  try:
    return  Pair([X.columns[0],y.columns[0]],1 - spatial.distance.cosine(X, y)) 
  except Exception as e: print(e)
```

```python
from datetime import datetime

def MyLog(message):
  now = datetime.now()
#   print(message,' date: ',now.today().strftime("%d/%m/%Y %H:%M:%S"))
```

**Information gain - mutual information In Classification**

Difference Between Information Gain And Mutual Information
I(X ; Y) = H(X) – H(X | Y) and IG(S, a) = H(S) – H(S | a)

As such, mutual information is sometimes used as a synonym for information gain. Technically, they calculate the same quantity if applied to the same data.

Recordemos que un evento más probable tiene menos contenido de información (no sorprendente), y un evento menos probable tiene más contenido de información (sorprendente).

```python

def Ig(X,y,name):
  MyLog('star IG')
  try:
    mutual_info = mutual_info_classif(X, y)
    return pd.DataFrame(mutual_info,
                    index=X.columns,
                    columns=["Score_Ig_"+name])
    MyLog('end IG')
  except:
     MyLog('error IG')
     return pd.DataFrame([],
                    index=X.columns,
                    columns=["Score_Ig_"+name])
 
```

```python

def ChiSquare(X,y,name):
  MyLog('star ChiSquare')
  try:
    chi2_selector = SelectKBest(chi2, k=5)
    chi2_selector.fit(X, y)
    # print(chi2_scores.sort_values(by=['score']))
    # kbest = np.asarray(X.columns)[chi2_selector.get_support()]
    # print(kbest)
    MyLog('end ChiSquare')
    return pd.DataFrame(chi2_selector.scores_,
                    index=X.columns,
                    columns=["Score_Chi_"+name])
  except:
     MyLog('error ChiSquare')
     return pd.DataFrame([],
                    index=X.columns,
                    columns=["Score_Chi_"+name])
  
```

```python
from ReliefF import ReliefF

def MyReliefF(X,y,n,name):
  MyLog('star MyReliefF')
  try:
    X_array,y_array = np.array(X),np.array(y)
    fs = ReliefF(n_neighbors=20, n_features_to_keep=n)
    result = fs.fit_transform(X_array, y_array)
    MyLog('end MyReliefF')
    return pd.DataFrame(np.asarray(fs.feature_scores)[fs.top_features],
                    index=np.asarray(X.columns)[fs.top_features],
                    columns=["Score_Rf_"+name])
  except:
     MyLog('error MyReliefF')
     return pd.DataFrame([],
                    index=X.columns,
                    columns=["Score_Rf_"+name])
```

```python

def find_indices(lst,a,b):
   for i, elem in enumerate(lst):
    if a in elem.pairList and b in elem.pairList:
     return elem
```

```python

def NTC(b,x,measure,column):
  MyLog('star NTC')
  similaritys = []
  # print('Inicio Similitud')
  for i, value in enumerate(x.columns):
    for j in range(i + 1, len(x.columns)):
      if measure == 'su':
        similaritys.append(SuPair(x.loc[:, [value]],x.loc[:, [x.columns[j]]])) 
      else:
        similaritys.append(CosPair(x.loc[:, [value]],x.loc[:, [x.columns[j]]]))
  c = []
  for i, valuei in enumerate(x.columns):
    feature = Features(valuei,[],[])
    c.append(feature)
    for j, valuej in enumerate(x.columns):
      if i != j:
          pair = find_indices(similaritys, valuei, valuej)
          if pair.value >= b:
            if valuej not in c[i].nameJ :
              c[i].nameJ.append(valuej)
              c[i].pairs.append(pair)
  import statistics
  result = []
  c = [i for i in c if len(i.pairs) > 0]
  while len(c) != 0:
    max = 0
    pos = -1
    for ci in c:
      mean = statistics.mean( [pair.value for pair in ci.pairs] )
      if mean > max:
        max = mean
        pos = ci
    if pos != -1:
      result.append(pos.name)
      c.remove(pos)
  # print('Termino de seleccion')
  MyLog('end NTC')
  return pd.DataFrame(['1' for i in result],
            index=result,
            columns=[column])
```

```python
import os

def FileList(directory,list=[]):
  for filename in os.listdir(directory):
      f = os.path.join(directory, filename)
      if os.path.isdir(f):
        FileList(f,list)
      # checking if it is a file
      if os.path.isfile(f) and os.path.splitext(f)[-1].lower() == '.csv':
        list.append(f)
  return np.array(list)
```

```python
import math

def CalculateRanking(total):
  return math.floor(
      math.fabs(
          math.sqrt(total) * math.log10(total)
      )
      ) 
```

```python

def GetName(file_location):
  file_name = os.path.basename(file_location)  
  head, tail = os.path.split(os.path.dirname(file_location))
  return tail +"_"+ file_name.replace(".csv", "") 
```

```python

def GetResults(path,level,b,rank):

    file_name = GetName(path)
    print(file_name)
    data=pd.read_csv(path)
    try:
      data=data.drop(columns=['Project','Hash', 'LongName','Parent'],axis=1, errors='ignore')
      X =data.drop(labels=['Number of Bugs'], axis=1).replace(np.nan,0).astype(float)
      y =  [1 if i > 0 else 0 for i in data['Number of Bugs']]
    except Exception as e:
        str(e)
#       print("Error in " + path + str(e))

    try:

      
      results = pd.DataFrame([],index=X.columns)
      metrics = pd.DataFrame(['1' for i in X.columns],
            index=X.columns,
            columns=[level])
      rank = (int)(len(X.columns) * rank)
#       print("Select rank",rank, " de ", len(X.columns))
      square = ChiSquare(X,y,file_name)
      square = square.sort_values(by=[square.columns[0]]).head(rank)
      ig = Ig(X,y,file_name)
      ig = ig.sort_values(by=[ig.columns[0]]).head(rank)
      Relief = MyReliefF(X,y,5,file_name)
      Relief = Relief.sort_values(by=[Relief.columns[0]]).head(rank)
      
 
      suSquare = NTC(b,X.loc[:,square.index],'su','CS_SU')
      suIg = NTC(b,X.loc[:,ig.index],'su','IG_SU')
      suRelief = NTC(b,X.loc[:,Relief.index],'su','RF_SU')

      cosSquare = NTC(b,X.loc[:,square.index],'cos','CS_COS')
      cosIg = NTC(b,X.loc[:,ig.index],'cos','IG_COS')
      cosRelief = NTC(b,X.loc[:,Relief.index],'cos','RF_COS')
      
      # print(df)
      # data = pd.concat([suSquare,suIg],axis=0)
    except Exception as e:
        str(e)
    return pd.concat([results,metrics,suSquare,suIg,suRelief,cosSquare,cosIg,cosRelief],axis=1)
  
```
